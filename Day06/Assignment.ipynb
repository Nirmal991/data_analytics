#### Assignment 1 
A temperature sensor sends continuous readings between $20.0^\circ\text{C}$ and $25.0^\circ\text{C}$.
- Generate 1,000 synthetic sensor readings
- Compute the mean and median of your generated dataset.
- Generate a Graphical Visualization.

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)

temperatures = np.random.uniform(low=20.0, high=25.0, size=1000)

mean_temp = np.mean(temperatures)
median_temp = np.median(temperatures)


print("Total readings:", len(temperatures))
print(f"Mean Temperature: {mean_temp:.2f}°C")
print(f"Median Temperature: {median_temp:.2f}°C")

plt.figure(figsize=(10,5))

plt.hist(temperatures, bins=30, edgecolor="black", alpha=0.7)

plt.axvline(mean_temp, linestyle="--", linewidth=2,
            label=f"Mean: {mean_temp:.2f}°C")

plt.axvline(median_temp, linestyle="-.", linewidth=2,
            label=f"Median: {median_temp:.2f}°C")

plt.xlabel("Temperature (°C)")
plt.ylabel("Frequency")
plt.title("Distribution of 1000 Synthetic Temperature Readings")
plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.show()
```

#### Assigment 2 
#### E-Commerce / Web Analytics (A/B Testing Simulator)
- Task:Write a scriptt to simulate 1,000 incoming website visitors.
- Assign each visitor a random float between 0.0 and 1.0.
- If the value is $< 0.5$, route them to 'Design A'. 
- Otherwise, route them to 'Design B'.
- Print the percentage of visitors assigned to each design. 

```python
np.random.seed(42)
num_visitors = 1000

random_values = np.random.uniform(0.0, 1.0, num_visitors)

design_a = np.sum(random_values < 0.5)
design_b = np.sum(random_values >= 0.5)

percentage_a = (design_a / num_visitors) * 100
percentage_b = (design_b / num_visitors) * 100

print("Total Visitors:", num_visitors)

print(f"Design A Visitors: {design_a}")
print(f"Design A Percentage: {percentage_a:.2f}%")

print(f"\nDesign B Visitors: {design_b}")
print(f"Design B Percentage: {percentage_b:.2f}%")
```

####  Assingment 4: E-Commerce Delivery Time Estimator
**Scenario:** An e-commerce platform estimates order delivery times. On average, a delivery takes **30 minutes** with a standard deviation of **5 minutes**.

**Tasks:**
1. Simulate delivery times for **1,000 orders**.
2. Calculate what percentage of deliveries took **longer than 40 minutes**.
3. Plot the delivery times and draw a threshold at the 40-minute mark.

```python
np.random.seed(42)

delivery_times = np.random.normal(
    loc=30,
    scale=5,
    size=1000
)

threshold = 40

late_deliveries = np.sum(delivery_times > threshold)

percentage_late = (late_deliveries/1000)* 100

print("Deliveries taking longer than 40 minutes:", late_deliveries) 
print(f"Percentage of late deliveries: {percentage_late:.2f}%")

plt.figure(figsize=(10, 5)) 
plt.hist( delivery_times, bins=30, edgecolor="black", alpha=0.7 ) 
plt.axvline( threshold, linestyle="--", linewidth=2, label="Threshold = 40 minutes" ) 
plt.xlabel("Delivery Time (minutes)") 
plt.ylabel("Number of Orders") 
plt.title("Distribution of Delivery Times for 1,000 Orders") 
plt.legend() 
plt.grid(axis="y", alpha=0.3) 
plt.show()
```

####  Assingment 5: Quality Control Bottling Audit
**Scenario:** A beverage factory fills soda bottles with a target volume of **500 mL** and a standard deviation of **2 mL**.

**Tasks:**
1. Generate **2,000 bottle volumes** .
2. Any bottle under **495 mL** is underfilled and rejected. Count how many bottles get rejected and calculate the rejection rate percentage.
3. Calculate the sample `mean` and `std` of your dataset to verify they match the factory setup.

```python
np.random.seed(42)

target_volume = 500 
population_std = 2 
num_bottles = 2000 
rejection_limit = 495

bottle_volumes = np.random.normal( loc=target_volume, scale=population_std, size=num_bottles )

rejected_bottles = np.sum(bottle_volumes < rejection_limit)

rejection_rate = (rejected_bottles / num_bottles) * 100

sample_mean = np.mean(bottle_volumes) 
sample_std = np.std(bottle_volumes, ddof=1)

print("Total bottles:", num_bottles) 
print("Target volume:", target_volume, "mL") 
print("Rejection limit:", rejection_limit, "mL") 
print("\n--- Quality Control Results ---") 
print("Rejected bottles:", rejected_bottles) 
print(f"Rejection rate: {rejection_rate:.2f}%") 
print("\n--- Statistical Verification ---") 
print(f"Target mean: {target_volume:.2f} mL") 
print(f"Sample mean: {sample_mean:.2f} mL") 
print(f"Target standard deviation: {population_std:.2f} mL") 
print(f"Sample standard deviation: {sample_std:.2f} mL")
```
